# Projeto LangChain + Ollama
## Desenvolvimento Local com Docker Desktop

Este notebook é para testar e desenvolver a aplicação com LangChain e Ollama localmente.


````python
# Importar Bibliotecas Necessárias
import os
from dotenv import load_dotenv
import requests
from langchain_community.llms import Ollama

# Carregar variáveis de ambiente
load_dotenv()

print("✅ Bibliotecas importadas com sucesso!")
````

## 1. Configurare Conexão com Ollama

Verificar se o Ollama está rodando em Docker e configurar a conexão.

````python
# Verificar conexão com Ollama
OLLAMA_BASE_URL = "http://localhost:11434"

try:
    response = requests.get(f"{OLLAMA_BASE_URL}/api/tags")
    if response.status_code == 200:
        modelos = response.json()
        print("✅ Conexão com Ollama estabelecida!")
        print(f"\n📦 Modelos disponíveis:")
        for modelo in modelos.get('models', []):
            print(f"  - {modelo['name']}")
    else:
        print("❌ Erro ao conectar com Ollama")
except requests.exceptions.ConnectionError:
    print("❌ Não foi possível conectar ao Ollama. Certifique-se que está rodando em Docker")
````

````python
# Inicializar LLM com Ollama
# Usando o modelo llama3.2:1b que foi baixado
llm = Ollama(
    base_url=OLLAMA_BASE_URL,
    model="llama3.2:1b"
)

print("✅ LLM Ollama inicializado!")
print(f"Modelo: llama3.2:1b")
````

## 2. Teste Simples do Modelo

Executar um prompt simples para testar se o modelo está funcionando corretamente.

````python
# Teste Simples: Fazer uma pergunta ao modelo
prompt = "Qual é a capital da Brasil em 1580?"
print(f"📝 Pergunta: {prompt}\n")
print("🤖 Resposta:\n")
resposta = llm.invoke(prompt)
print(resposta)
print(f"\n✅ Teste concluído!")
````

## 3. Uso de PromptTemplates

Criar templates reutilizáveis para diferentes tipos de prompts.

````python
from langchain_core.prompts import PromptTemplate

# Criar um template customizado para resumir texto
template_resumo = PromptTemplate(
    input_variables=["texto"],
    template="""Você é um assistente útil. 
    
Favor resumir o seguinte texto em 2-3 sentenças:

{texto}

Resumo:"""
)

# Testar o template
texto_exemplo = "Python é uma linguagem de programação de alto nível, interpretada e com tipagem dinâmica. É muito utilizada em desenvolvimento web, análise de dados, inteligência artificial e automação."

prompt_formatado = template_resumo.format(texto=texto_exemplo)
print(f"📝 Texto original:\n{texto_exemplo}\n")
print("Gerando resumo...")
print("\n🤖 Resumo:\n")
resultado = llm.invoke(prompt_formatado)
print(resultado)
````

## 4. Usando LangChain Chains

Criar chains para processar dados através de múltiplas etapas.

````python
from langchain_core.runnables import RunnablePassthrough
from langchain_core.prompts import PromptTemplate

# Criar um template para um assistente de programação
template_programacao = PromptTemplate(
    input_variables=["linguagem", "problema"],
    template="""Você é um especialista em {linguagem}.

Problema: {problema}

Forneça uma solução clara e bem estruturada com exemplos de código."""
)

# Criar a chain usando o operador pipe |
chain_programacao = template_programacao | llm

# Testar a chain
linguagem = "Python"
problema = "Como criar uma função que verifica se um número é primo?"

print(f"💻 Linguagem: {linguagem}")
print(f"❓ Problema: {problema}\n")
print("🤖 Solução:\n")
resposta_chain = chain_programacao.invoke({
    "linguagem": linguagem,
    "problema": problema
})
print(resposta_chain)
````
